```python
import os

def folder_has_any_file(folder_path):
    return any(
        os.path.isfile(os.path.join(folder_path, f)) and not f.startswith(".")
        for f in os.listdir(folder_path)
    )

folder_path = "../../docs"
if folder_has_any_file("../../docs"):
    print("yes")
else:
    print("no")
```
